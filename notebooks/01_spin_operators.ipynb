# Spin operators and two-spin Hilbert spaces

This notebook develops the spin-1/2 operators required to reproduce
selected results from Jo & Kim (2022), "Simulating open quantum many-body
systems using optimised circuits in digital quantum simulation."

The first goal is to construct the Pauli operators and extend them to a
two-spin Hilbert space using tensor products.

```python
import numpy as np

sigma_x = np.array([
    [0, 1],
    [1, 0]
], dtype=complex)

sigma_y = np.array([
    [0, -1j],
    [1j, 0]
], dtype=complex)

sigma_z = np.array([
    [1, 0],
    [0, -1]
], dtype=complex)

I = np.eye(2, dtype=complex)
```

```python
print(sigma_x @ sigma_x)
print(sigma_y @ sigma_y)
print(sigma_z @ sigma_z)
```

```python
print("σx² = I:", np.allclose(sigma_x @ sigma_x, I))
print("σy² = I:", np.allclose(sigma_y @ sigma_y, I))
print("σz² = I:", np.allclose(sigma_z @ sigma_z, I))
```

```python
sx1 = np.kron(sigma_x, I)
sx2 = np.kron(I, sigma_x)

sy1 = np.kron(sigma_y, I)
sy2 = np.kron(I, sigma_y)

sz1 = np.kron(sigma_z, I)
sz2 = np.kron(I, sigma_z)
```

```python
print(sz1)
```

```python
J = 1.0
Delta = 1.0

H_interaction = -J * (sz1 @ sz2)

H_field = Delta * (sx1 + sx2)

H = H_interaction + H_field

print(H)
```

```python
eigenvalues, eigenvectors = np.linalg.eigh(H)

print("Eigenvalues:")
print(eigenvalues)
```

```python
up = np.array([1, 0], dtype=complex)
down = np.array([0, 1], dtype=complex)

psi0 = np.kron(up, up)

print(psi0)
```

```python
norm = np.vdot(psi0, psi0)

print(norm)
```

```python
Mz = (sz1 + sz2) / 2

expectation_Mz = np.vdot(psi0, Mz @ psi0)

print("Initial magnetisation <Mz> =", expectation_Mz)
```

```python
# Construct the four two-spin states
states = {
    "up-up": np.kron(up, up),
    "up-down": np.kron(up, down),
    "down-up": np.kron(down, up),
    "down-down": np.kron(down, down)
}

# Use the Mz operator you have already constructed
for name, psi in states.items():
    expectation = np.vdot(psi, Mz @ psi)
    print(f"{name}: <Mz> = {expectation.real:.1f}")
```
